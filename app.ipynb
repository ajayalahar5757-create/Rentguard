```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
import re
import joblib
import string
```

```python
fake=pd.read_csv(r'C:\Users\saini\Downloads\fake_rental_same_format_100x12 (1).csv')
true=pd.read_csv(r'C:\Users\saini\Downloads\real_rental_dataset_100_rows (1).csv')
```

```python
fake.head()
```

```python
true.head()
```

```python
fake["class"] = 0
true["class"] = 1
```

```python
data = pd.concat([fake, true], axis=0)
```

```python
data.sample(10)
```

```python
data = data.drop(['Posted On', 'Floor', 'Area Type', 'Furnishing Status', 'Bathroom', 'Tenant Preferred'], axis=1)
```

```python
data.sample(5)
```

```python
x=data[['BHK','Rent','Size','Area Locality','City','Point of Contact']].fillna('').agg(' '.join, axis=1)
y=data['class']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
vectorizer = TfidfVectorizer()
xv_train = vectorizer.fit_transform(x_train)
xv_test = vectorizer.transform(x_test)
```

```python
lr= LogisticRegression()
lr.fit(xv_train, y_train)
```

```python
predictions = lr.predict(xv_test)
lr.score(xv_test, y_test)
```

```python
print(classification_report(y_test, predictions))
```

```python
joblib.dump(vectorizer, 'vectorizer.jb')
joblib.dump(lr, 'lr_model.jb')
```
